```python
dataset=[]
```

```python
label=[]
```

```python
for f_name in fearful:
    f_image=cv2.imread(path_fearful+f_name)
    f_image=cv2.resize(f_image,(48,48))
    f_image=np.array(f_image)
    dataset.append(f_image)
    label.append(0)
    
```

```python
for a_name in angry:
    a_image=cv2.imread(path_angry+a_name)
    a_image=cv2.resize(a_image,(48,48))
    a_image=np.array(a_image)
    dataset.append(a_image)
    label.append(1)
```

```python
for h_name in happy:
    h_image=cv2.imread(path_happy+h_name)
    h_image=cv2.resize(h_image,(48,48))
    h_image=np.array(h_image)
    dataset.append(h_image)
    label.append(2)
```

```python
for n_name in neutral:
    n_image=cv2.imread(path_neutral+n_name)
    n_image=cv2.resize(n_image,(48,48))
    n_image=np.array(n_image)
    dataset.append(n_image)
    label.append(3)
```

```python
for s_name in sad:
    s_image=cv2.imread(path_sad+s_name)
    s_image=cv2.resize(s_image,(48,48))
    s_image=np.array(s_image)
    dataset.append(s_image)
    label.append(4)
```

```python
for s_name in surprised:
    s_image=cv2.imread(path_surprised+s_name)
    s_image=cv2.resize(s_image,(48,48))
    s_image=np.array(s_image)
    dataset.append(s_image)
    label.append(5)
```

```python
ary_dataset=np.array(dataset)
```

```python
ary_label=np.array(label)
```

```python
X=ary_dataset.copy()
```

```python
y=ary_label.copy()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
X_train_scale= X_train / 255.0
X_test_scale = X_test / 255.0
```

```python
X_train_scale.max()
```

```python
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras.layers import Input,Dense,Conv2D,MaxPooling2D,Flatten,Dropout,BatchNormalization
```

```python
cnn_model=Sequential()
```

```python
cnn_model.add(Input(shape=(48,48,3)))
```

```python
cnn_model.add(Conv2D(filters=32,kernel_size=(3,3),strides=(1,1),padding="valid",activation="relu"))
```

```python
cnn_model.add(MaxPooling2D(pool_size=(2,2),padding="valid"))
```

```python
cnn_model.add(Conv2D(64,kernel_size=(3,3),strides=(1,1),padding="valid",activation="relu"))
```

```python
cnn_model.add(MaxPooling2D(pool_size=(2,2),padding="valid"))
```

```python
cnn_model.add(Conv2D(128,kernel_size=(3,3),strides=(1,1),padding="valid",activation="relu"))
```

```python
cnn_model.add(MaxPooling2D(pool_size=(2,2),padding="valid"))
```

```python
cnn_model.add(Flatten())
```

```python
cnn_model.add(Dense(128,activation="relu"))
```

```python
cnn_model.add(Dense(64,activation="relu"))
```

```python
cnn_model.add(Dense(6,activation="softmax"))
```

```python
cnn_model.summary()
```

```python
cnn_model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
```

```python
cnn_model.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=10,
    batch_size=32
)
```

```python
cnn_model.evaluate(X_test, y_test)
```
