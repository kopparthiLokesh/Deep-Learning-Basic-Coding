## porter stemmer

```python
import nltk
```

```python
from nltk.stem import PorterStemmer
```

```python
pstem=PorterStemmer()
```

```python
len(words)
```

```python
import time
```

```python
for w in words:
    print("Original :",w)
    print("porter :",pstem.stem(w))
    print("--------")
    print()
    time.sleep(1)
```

## Snow Ball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snow_stemmer=SnowballStemmer(language="english")
```

```python
snow_stemmer
```

```python
snow_stemmer.stem("working")
```

```python
len(words)
```

```python
for w in words:
    print("Original :",w)
    print("porter :",pstem.stem(w))
    print("Snowball :",snow_stemmer.stem(w))
    print("-----------------")
    print()
    time.sleep(1)
```

## Lancaster stemmer

```python
from nltk.stem import LancasterStemmer
```

```python
lan=LancasterStemmer()
```

```python
lan.stem("writes")
```

```python
for w in words:
    print("Original :",w)
    print("porter :",pstem.stem(w))
    print("Snowball :",snow_stemmer.stem(w))
    print("Lancaster :",lan.stem(w))
    print("-----------------")
    print()
    time.sleep(1)
```

## lemmatization

```python
import nltk
```

```python
nltk.download("wordnet")
```

```python
lemma=nltk.WordNetLemmatizer()
```

```python
lemma.lemmatize("words")
```

```python
lemma.lemmatize("working")
```

```python
for w in words:
    print("Original :",w)
    print("porter :",pstem.stem(w))
    print("Snowball :",snow_stemmer.stem(w))
    print("Lancaster :",lan.stem(w))
    print("Lemmatize :",lemma.lemmatize(w))
    print("-----------------")
    print()
    time.sleep(1)
```
